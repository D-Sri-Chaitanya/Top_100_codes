```python
#patterns:
# square star pattern:
n = int(input())
for i in range(n):
    for j in range(n):
        print("*", end = " ")
    print()
```

```python
# Hollow square pattern:
n = int(input("Enter size: "))
for i in range(1, n+1):          
    for j in range(1, n+1):     
        
        if i == 1 or i == n or j == 1 or j == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
            
    print()
```

```python
#Rhombus Star Pattern:
n = int(input("Enter number of rows:"))
for i in range(0,n):
    for j in range(0,i):
        print(" ",end="")
    for j in range(0,n):
        print("*",end = "")
    print()
```

```python
#Rectangle Star Pattern:
n = int(input("Enter number of rows: "))
m = int(input("Enter number of columns: "))
for i in range(0,n):
    for j in range(0,m):
        print("*",end = "")
    print()
```

```python
#Hollow Rectangle Star Pattern:
n = int(input("Enter number of rows: "))
m = int(input("Enter number of columns: "))
for i in range(0,n):
    for j in range(0,m):
        if i == 0 or j == 0 or i == n-1 or j == m-1:
            print("*",end="")
        else:
            print(" ",end="")
    print()
```
